## Part 2: Find Researchers using the openAlex API

### Steps
1. Retrive Data

```python
# Import unique organizer names found in the IC2S2 conference in 2025 (originally found by web scraping)
organizer_names_df = pd.read_csv("./data/organizer_names.csv")
organizer_names_df
```

```python
#Setup configuration for openalexnet API endpoint for data retrievel
BASE_URL = "https://api.openalex.org/"
RESOURCE = "authors"
url = BASE_URL+RESOURCE
url
```

```python
#This function gets applied on all organizer names to find the desired fields
def get_author(name, API_KEY):

    local_params = {"search": name, "api_key": API_KEY}
    response = requests.get(url, local_params)
    response_json = response.json()

    if "results" in response_json and len(response_json["results"]) > 0: # Check if the person is in the database
        result = response_json["results"][0]
        try: # Throws an error, when the country code property cant be accessed
            country_code = result["last_known_institutions"][0]["country_code"]  
            return{
            "id": result["id"],
            "display_name": result["display_name"],
            "works_api_url": result["works_api_url"],
            "h_index": result["summary_stats"]["h_index"],
            "works_count": result["works_count"],
            "country_code": country_code
            }
        
        except: # Error: set country_code to NaN
            return{
            "id": result["id"],
            "display_name": result["display_name"],
            "works_api_url": result["works_api_url"],
            "h_index": result["summary_stats"]["h_index"],
            "works_count": result["works_count"],
            "country_code": np.nan
            }
            
    
    else: # Name does not exist in the data base. Set all fields to NaN
            return{
            "id": np.nan,
            "display_name": np.nan,
            "works_api_url": np.nan,
            "h_index": np.nan,
            "works_count": np.nan,
            "country_code": np.nan
            }
```

```python
#Applies the get_author function to all organizer_names_df names in parralel using the ThreadPoolExecutor package.
API_KEY = os.getenv('JULIE_API_KEY_2')
with ThreadPoolExecutor(max_workers=10) as executor:
    results1 = list(executor.map(partial(get_author, API_KEY=API_KEY), list(organizer_names_df["Name"])[:790]))

API_KEY = os.getenv('JULIE_API_KEY_4')
with ThreadPoolExecutor(max_workers=10) as executor:
    results2 = list(executor.map(partial(get_author, API_KEY=API_KEY), list(organizer_names_df["Name"])[790:]))
```

2. Data storage

```python
# We store the results in a datafram and filter out the authors who do not have a working api url.
results = results1+results2
metadata_df = pd.DataFrame(results)
filtered_metadata_df = metadata_df[metadata_df["works_api_url"].notna()]
filtered_metadata_df
```

```python
# The panda df can then be stored in a file
filtered_metadata_df.to_csv(r"data/D1.csv", index=False)
```

## Part 3: Collect Research Articles



### Steps

1. Retrieve Data

```python
# Reads D1 from the file saved just for fun and filters out authors without specified amount of works

# Filter
# Only include *IC2S2 authors* with a total work count between 5 and 5,000
D1 = pd.read_csv(r"data/D1.csv")
D1 = D1[(D1["works_count"] >= 5) & (D1["works_count"] <= 5000)].reset_index(drop=True)
D1
```

```python
# Helper function for doing bulk request and getting the author_ID from openAlex_ID

def get_id(openalex_id):
    return openalex_id.split("/")[-1]

def chunked(authors, size):
    for i in range(0, len(authors), size):
        yield authors[i : i + size]

def build_works_url(author_batch):
    author_filter = "|".join(author_batch)
    return (
        f"https://api.openalex.org/works?filter=authorships.author.id:{author_filter}"
    )

# Applies the helper function to get the author batches.
original_authors_id = D1["id"].apply(get_id)
author_batches = list(chunked(original_authors_id, 25))
```

```python
# Setup filters for the specified concepts
SOCIAL_CONCEPTS = {"Sociology", "Psychology", "Economics", "Political science"}
QUANT_CONCEPTS = {"Mathematics", "Physics", "Computer science"}


def is_relevant_concepts(article):
    concepts = article["concepts"]
    social = any(
        c["level"] == 0 and c["display_name"] in SOCIAL_CONCEPTS for c in concepts
    )
    quant = any(
        c["level"] == 0 and c["display_name"] in QUANT_CONCEPTS for c in concepts
    )
    return social and quant
```

```python
#Get the work by authors in batches
def get_response(works_api_url):
    rows = []
    params = {"per_page": 200, "api_key": API_KEY, "cursor": "*"}

    session = requests.Session()
    session.headers.update({"User-Agent": "research-script/1.0"})

    while True:
        response = session.get(works_api_url, params=params, timeout=30)
        result = response.json()
        results = result["results"]

        for article in results:
            # Concepts filter
            if not is_relevant_concepts(article):
                continue

            authors = [
                authorship["author"]["id"] for authorship in article["authorships"]
            ]

            # Filter
            # Retrieve only works that have received more than 10 citations.
            # Limit to works authored by fewer than 10 individuals.
            if article["cited_by_count"] <= 10 or len(authors) >= 10:
                continue

            rows.append(
                {
                    "id": article["id"],
                    "publication_year": article["publication_year"],
                    "cited_by_count": article["cited_by_count"],
                    "author_ids": [
                        author.split("/")[-1]
                        for author in authors
                        if author is not None
                    ],
                    "title": article["title"],
                    "abstract_inverted_index": article["abstract_inverted_index"],
                }
            )

        next_cursor = result["meta"]["next_cursor"]
        if not next_cursor:
            break
        params["cursor"] = next_cursor

    return pd.DataFrame(rows)
```

```python
# Runs the code in parrallel using the Parallel library and the tqdm package to see the process.

urls = [build_works_url(batch) for batch in author_batches]

metadata_list = Parallel(n_jobs=4)(
    delayed(get_response)(url) for url in tqdm(urls, desc="Fetching works (batched)")
)

# Filter out None results and duplicates and combine
metadata_list = [df for df in metadata_list if df is not None and not df.empty]
metadata_df = pd.concat(metadata_list, ignore_index=True)
metadata_df = metadata_df.drop_duplicates(subset=["id"])
```

2. Data storage

```python
# Splits up the metadata and stores it in csv files
D2 = metadata_df[["id", "publication_year", "cited_by_count", "author_ids"]].copy()
D3 = metadata_df[["id", "title", "abstract_inverted_index"]].copy()

D2.to_csv("./data/D2.csv", index=False)
D3.to_csv("./data/D3.csv", index=False)
```
