```python
import pandas as pd
import numpy as np
import requests
```

```python
df = pd.read_csv("organizer_names.csv")
df
```

```python
BASE_URL = "https://api.openalex.org/"
RESOURCE = "authors"
complete_url = BASE_URL + RESOURCE
complete_url
```

```python
def get_response(name):
    url = "https://api.openalex.org/authors"
    response = requests.get(
        url, params={"search": name, "api_key": "7G7kFeCnnbewza7W5xWqg3"}
    )
    result = response.json()
    result = result["results"]
    print(name)

    if len(result) > 0:  # Tjek om personen findes i databasen
        result = result[0]
        try:  # Error nogle gange, når denn ikke kan blive accessed
            dummy = result["last_known_institutions"][0]["country_code"]
        except:  # Error
            return {
                "id": result["id"],
                "display_name": result["display_name"],
                "works_api_url": result["works_api_url"],
                "h_index": result["summary_stats"]["h_index"],
                "works_count": result["works_count"],
                "country_code": np.nan,
            }
        else:  # Ingen problemer
            return {
                "id": result["id"],
                "display_name": result["display_name"],
                "works_api_url": result["works_api_url"],
                "h_index": result["summary_stats"]["h_index"],
                "works_count": result["works_count"],
                "country_code": result["last_known_institutions"][0]["country_code"],
            }

    else:  # Navnet findes ikke i databasen
        return {
            "id": np.nan,
            "display_name": np.nan,
            "works_api_url": np.nan,
            "h_index": np.nan,
            "works_count": np.nan,
            "country_code": np.nan,
        }
```

```python
metadata_df = df["Name"].apply(get_response).apply(pd.Series)
```
